```python
import numpy as np

# Daily step counts: morning and evening for 5 days
steps = np.array([
    [2000, 3000],
    [2500, 3500],
    [1800, 3200],
    [2200, 4000],
    [2800, 4500]
])
original_shape = steps.shape

steps_1d = steps.reshape(10)

shape_1d = steps_1d.shape
steps_2d = steps_1d.reshape(5, 2)
shape_2d = steps_2d.shape

original_shape, steps_1d, shape_1d, steps_2d, shape_2d
```

```python
import numpy as np

# Prices of 12 food items
prices = np.array([
    120, 150, 200, 90,
    180, 250, 110, 160,
    300, 140, 220, 100
])

# Convert into a 3x4 array
prices_2d = prices.reshape(3, 4)

# ravel() returns a flattened view when possible
# Changes to the returned array may affect the original array
raveled_prices = prices_2d.ravel()

# flatten() always returns a copy
# Changes to the returned array do not affect the original array
flattened_prices = prices_2d.flatten()

# resize() changes the shape of the original array
# Here, the original array becomes 4x3
prices_2d.resize(4, 3)

prices_2d, raveled_prices, flattened_prices
```

```python
import numpy as np

# Rows represent playlists; columns represent Pop, Rock, and Indie
playlist = np.array([
    [10, 5, 8],
    [7, 12, 4],
    [15, 6, 9]
])

# Transpose using the T attribute
playlist_T = playlist.T

# Transpose using np.transpose()
playlist_transposed = np.transpose(playlist)

playlist_T, playlist_transposed
```

```python
import numpy as np

# Ratings of 15 Flipkart products
ratings = np.array([
    4, 5, 3, 4, 2,
    5, 4, 3, 5, 4,
    2, 3, 4, 5, 1
])

# Convert 1D array into a 3x5 array
ratings_2d = ratings.reshape(3, 5)

# Convert the 2D array back into a 1D array
ratings_1d = ratings_2d.flatten()

# Use flatten() when an independent copy is needed.
# Use ravel() when a view is acceptable and avoiding a copy is useful.
ratings_2d, ratings_1d
```
